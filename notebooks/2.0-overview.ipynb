# 2.0 Overview of the clean data

The classic first look at a dataset: shape, columns, missing values, categories.
The full table is loaded with Polars (about 4 GB in memory). For pandas-style exploration we use a random sample.

```python
import polars as pl

from realtime_recsys.config import PROCESSED_DATA_DIR

df = pl.read_parquet(PROCESSED_DATA_DIR / "events_clean.parquet")
df.shape
```

## First rows and column types

pandas: `df.head()` and `df.dtypes`. Polars: `df.head()` and `df.schema`.

```python
df.head()
```

```python
df.schema
```

## Missing values

pandas: `df.isna().sum()`. Polars: `df.null_count()`.

```python
df.null_count()
```

## Event types

pandas: `df["event_type"].value_counts()`. Polars: same name.

```python
df["event_type"].value_counts()
```

## Unique values per column

pandas: `df.nunique()`. Polars: `df.select(pl.all().n_unique())`.

```python
df.select(pl.all().n_unique())
```

## pandas on a random sample

One million random rows converted to a pandas DataFrame. Use it for quick pandas-style checks.
Numbers from a sample are approximate; use the full table (above) for exact counts.

```python
sample = df.sample(1_000_000, seed=42).to_pandas()
sample.info()
```
